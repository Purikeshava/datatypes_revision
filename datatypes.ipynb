```python
#1.Numeric Data Types
#Numeric data types are used to store numbers.
#stores int, float, complex
a = 100 # integer
b = 1.1 # float
c = 3+4j # complex  here 3 is real part and 4j is imaginary part
```

```python
print(type(a))
print(type(b))
print(type(c))
```

```python
#2.Boolean Data Type (bool)
#Boolean has only two values:
# TRUE , FALSE 
# this data types are used to compare the values and find true or false
a = 10 
b = 20
print(a<b)
print(a>b)
# real world example
salary  = 10000
print (salary>50000)
```

```python
#3.String (str)
#A string is a collection of characters enclosed within quotes.
#We can use:
#Single quotes ' '
#Double quotes " "
#Triple quotes ''' '''
name1 = '  ironman' # single quotes
name2 = "spiderman" # double quotes
name3 = '''batman''' # triple quotes
print(type(name1))
print(type(name2))
print(type(name3))
```

```python
# we can do operations on the string data types like 
print(name1[4]) # to specify index numbering to find letter in words
print(name2[0:4]) # to specify sequence of letters from word
print(len(name3)) # to find the length of the word
print(name1.upper()) # to print letters in uppercase
print(name1.lower()) # to print letters in lowercase
print(name3.replace("batman","superman")) # to replace the words when u print
print(name1.strip()) #Removes leading and trailing whitespace. i kept space infront of ironman in variable
```
