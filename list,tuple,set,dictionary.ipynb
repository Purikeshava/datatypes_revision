```python
#2----------------tuple-----------------------
# ******** tuple  elements are stored in () (Round brackets)
# -----cannot change elements in the tuple called un-mutable
# -----duplicate elements can be stored  
# -----mainly used for storing fixed items in the list 
#------support indexing 
#------maintain order of elements in tuple
numbers = (10,20,30,40,50) 
print(numbers)
```

```python
# same we can use positive and negative indexing in tuple as used in list 
print(numbers[2]) # postive indexing 
print(numbers[-2]) # negative indexing 
# we can use all the functions used in list but we cannot add new elements into tuple its fixed length
```

```python
a = (10) # normal integer variable 
b = (20,) # in this we used coma , thats why it become tupple can see in output 
print(type(a))
print(type(b))
```

```python
## Tuples are useful for storing fixed values such as coordinates or a fixed set of configuration values
```

```python
#3----------------set-----------------------
# ******** set  elements are stored in {} (Flower brackets)
# -----can change elements in the set called mutable
# -----duplicate elements cannot be stored no duplicates  
# -----mainly used for storing unique items in the set 
#------no  indexing in set 
#------cannot maintain order of elements in set
#------Sets are mutable, but they do not support indexing or slicing.
sets = {10,20,30,40,50,50,20,60,10}
print(sets)
# you can see here order in input and output are different  and removed duplicates 
# due to in-consistent order indexing and slicing is not done in set but if you want convert to list 
```

```python
sets.add(100) # adding new element in set 
print(sets)  # added 100 
```

```python
sets.update([999]) # to update set
print(sets) # updated set inserted 999
```

```python
sets.remove(999) # to remove element in set 
print(sets) # removed 999
```

```python
# to perform set operations take two sets 
a = {10,20,30,40,50}
b = {60,70,80,10,20}
print(a)
print(b)
```

```python
print(a|b)  # use | to get unique values from both sets
```

```python
print(a&b) # use & to get common values from both sets 
```

```python
print(a-b) # use - to get unique values from second set means b 
print(b-a)  # use - to get unique values from second set means a
```

```python
print(a^b) # use ^ to get values that are in either set, but not in both.
```

```python
#4----------------dictionary-----------------------
# ******** dictionary  elements are stored in {} (Flower brackets)
# -----can change elements in the dictionary called mutable
# -----duplicate elements cannot be stored no duplicates  
# -----mainly used for storing key-value pair items in the dictionary 
#------indexing in set but access using keys 
#------ maintain order of elements in dictionary 
# A dictionary stores information in key-value pairs.
# Think of it as storing a person's details under labels.
person = { "name" : "prime",
           "age"  :  20,
           "branch" : "cse",
           "marks" : 99  }
print(person)
         
```

```python
# Here:
# "name" is a key "prime" is its value.
# "age" is a key 20 is its value.
# "branch" is a key "cse" is a value.
# "marks" is a key 99 is a value 
# indexing and slicing is done by using the keys 
```

```python
print(person.get("name")) # use get() and key "name" to retrive data 
print(person.get("branch"))
print(person.get ("city","not available")) # used to understand data in dictionary 
```

```python
print(person.keys()) # to know keys in dictionary
```

```python
print(person.values()) # to know values in dictionary 
```

```python
print(person.items()) # to know key value pairs 
```
